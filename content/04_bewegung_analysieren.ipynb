# Bewegungen analysieren

Bisher hast du Folgen berechnet. Jetzt wenden wir sie auf **Bewegungen** an.

Eine Messung liefert den Ort `s` zu den Zeitpunkten `t = 0, Δt, 2Δt, …`. Das ist eine Folge s(n).
Daraus lassen sich zwei neue Folgen bilden:

- **Geschwindigkeit:** v(n) = (s(n+1) − s(n)) / Δt  (Differenzenfolge von s, geteilt durch Δt)
- **Beschleunigung:** a(n) = (v(n+1) − v(n)) / Δt  (Differenzenfolge von v, geteilt durch Δt)

## Rechnen mit ganzen Folgen auf einmal

Eine Schleife berechnet ein Glied nach dem anderen. numpy kann auch mit **ganzen Folgen** auf einmal rechnen.
Das kennst du schon: `2 * t` oder `t**2` rechnet alle Glieder von `t` in einem Schritt.

Auch die Differenzenfolge geht so:

- `s[1:]` ist die Folge ohne das erste Glied: s(1), s(2), …
- `s[:-1]` ist die Folge ohne das letzte Glied: s(0), s(1), …
- `s[1:] - s[:-1]` ist dann s(n+1) − s(n), also die Differenzenfolge. Dafür gibt es die Kurzform `np.diff(s)`.
- Durch `dt` wird wieder jedes Glied geteilt.

Prüfe, dass du dasselbe Ergebnis wie mit der Schleife bekommst.
