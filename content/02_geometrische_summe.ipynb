# Summenfolge und geometrische Summe

Aus einer Folge a kann man die **Summenfolge** s bilden: man addiert die Glieder nacheinander auf.

s(0) = a(0),  s(n) = s(n−1) + a(n)

Beispiel: Ein Stück Schokolade. Du isst die Hälfte, dann die Hälfte vom Rest, dann wieder die Hälfte vom Rest …
Wie viel hast du insgesamt gegessen?

```python
import numpy as np
import matplotlib.pyplot as plt

N = 15
a = np.zeros(N)
s = np.zeros(N)

for n in range(N):
    a[n] = 0.5**(n+1)        # 1/2, 1/4, 1/8, ...

s[0] = a[0]
for n in range(1, N):
    s[n] = s[n-1] + a[n]     # Summenfolge

print(s)
plt.plot(a, "o", label="a(n)")
plt.plot(s, "o", label="s(n)")
plt.xlabel("n")
plt.legend()
plt.grid()
```

## Allgemeine geometrische Summe

Jetzt beginnen wir mit 1 und multiplizieren immer mit dem Faktor q:

1 + q + q² + q³ + …

```python
q = 0.5
N = 30
a = np.zeros(N)
s = np.zeros(N)

a[0] = 1
for n in range(1, N):
    a[n] = a[n-1] * q

s[0] = a[0]
for n in range(1, N):
    s[n] = s[n-1] + a[n]

print("letztes Glied der Summenfolge:", s[-1])
plt.plot(s, "o")
plt.xlabel("n")
plt.ylabel("s(n)")
plt.grid()
```
