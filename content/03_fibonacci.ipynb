# Die Fibonacci-Folge und der goldene Schnitt

Leonardo Fibonacci stellte 1202 die Kaninchen-Aufgabe:
Jedes Kaninchenpaar bekommt ab dem zweiten Lebensmonat jeden Monat ein neues Paar.
Wie viele Paare gibt es nach n Monaten?

Daraus ergibt sich die **Fibonacci-Folge**:

f(0) = 1,  f(1) = 1,  f(n) = f(n−1) + f(n−2)

Also: 1, 1, 2, 3, 5, 8, …
