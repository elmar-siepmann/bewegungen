# Die Fibonacci-Folge und der goldene Schnitt

Leonardo Fibonacci stellte 1202 die Kaninchen-Aufgabe:
Jedes Kaninchenpaar bekommt ab dem zweiten Lebensmonat jeden Monat ein neues Paar.
Wie viele Paare gibt es nach n Monaten?

Daraus ergibt sich die **Fibonacci-Folge**:

f(0) = 1,  f(1) = 1,  f(n) = f(n−1) + f(n−2)

Also: 1, 1, 2, 3, 5, 8, …

## Aufgabe 1: Programmiere die Fibonacci-Folge

Unten steht der Code einer rekursiven Folge wie in `01_folgen`.
Ändere ihn so ab, dass er die Fibonacci-Folge berechnet. Lass die ersten 20 Glieder ausgeben und zeichnen.

<details>

Wenn `f[0]` und `f[1]` vorgegeben sind, muss die Schleife bei n = 2 beginnen: `for n in range(2, N):`
</details>

```python
import numpy as np
import matplotlib.pyplot as plt

N = 20
f = np.zeros(N)
f[0] = 1

for n in range(1, N):
    f[n] = f[n-1] + 2      # hier muss die Fibonacci-Regel stehen

print(f)
plt.plot(f, "o")
plt.xlabel("n")
plt.ylabel("f(n)")
plt.grid()
```
