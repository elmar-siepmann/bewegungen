## Ausprobieren

Diese Zelle funktioniert auch allein: Sie simuliert eine Feder und spielt sie ab. Damit kannst du testen, ob die Animation bei dir läuft.

```python
m = 0.2
D = 5.0
dt = 0.05
N = 120

s = 0.1
v = 0.0
t_folge = [0.0]
s_folge = [s]

for n in range(N):
    a = -D/m * s
    v = v + a*dt
    s = s + v*dt
    t_folge.append((n+1) * dt)
    s_folge.append(s)

try:
    from animation import animiere
except ModuleNotFoundError:   # Notlösung (iPad): Datei direkt von der Webseite laden
    from pyodide.http import open_url
    exec(open_url("https://elmar-siepmann.github.io/bewegungen/files/animation.py").read())

animiere(t_folge, s_folge)
```
