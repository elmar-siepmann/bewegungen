# Folgen berechnen und zeichnen

Eine Folge ist eine Liste von Zahlen: a(0), a(1), a(2), …

Python speichert die Folge in einem **Array** `a`. Das Glied a(n) heißt in Python `a[n]`.
Achtung: Python fängt bei 0 an zu zählen!

## 1. Explizite Folge

Jedes Glied wird direkt aus n berechnet, hier a(n) = 7 − 2·n.

```python
import numpy as np
import matplotlib.pyplot as plt

N = 10              # Anzahl der Glieder
a = np.zeros(N)     # Array mit N Nullen

for n in range(N):
    a[n] = 7 - 2*n

print(a)
plt.plot(a, "o")
plt.xlabel("n")
plt.ylabel("a(n)")
plt.grid()
```

## 2. Rekursive Folge

Jedes Glied wird aus dem **vorherigen** Glied berechnet:

a(0) = 3,  a(n) = a(n−1) + 2

Das erste Glied muss man vorgeben, die Schleife beginnt deshalb erst bei n = 1.

```python
N = 10
a = np.zeros(N)
a[0] = 3

for n in range(1, N):
    a[n] = a[n-1] + 2

print(a)
plt.plot(a, "o")
plt.xlabel("n")
plt.ylabel("a(n)")
plt.grid()
```

**Aufgaben**

   Beschreibe jeweils, wie sich die Folge verhält.

   Ändere den Startwert a[0]. Ändert sich der Grenzwert?
3. Kannst du die Zahl aus Aufgabe 2 auch ohne Python vorhersagen?
   Tipp: Wenn sich nichts mehr ändert, gilt a(n) = a(n−1).

## 3. Mehrere Folgen in einem Diagramm

Mit `label` und `plt.legend()` kann man Folgen vergleichen.

```python
N = 15
a = np.zeros(N)
b = np.zeros(N)
a[0] = 1
b[0] = 1

for n in range(1, N):
    a[n] = a[n-1] + 3      # arithmetisch: immer + 3
    b[n] = b[n-1] * 1.3    # geometrisch: immer · 1,3

plt.plot(a, "o", label="a: + 3")
plt.plot(b, "o", label="b: · 1,3")
plt.xlabel("n")
plt.legend()
plt.grid()
```
