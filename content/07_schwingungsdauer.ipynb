## Aufgabe 1: Wann geht die Feder durch die Mitte?

Die Feder aus dem letzten Notebook: m = 0,2 kg, D = 5 N/m, Start bei s = 0,1 m in Ruhe. Jetzt mit kleinem Zeitschritt Δt = 0,01 s, insgesamt 6 Sekunden.

Neu ist die **if-Abfrage** am Ende der Schleife. Vor jedem neuen Ort merkt sich die Schleife den alten Ort in `s_alt`. Dann prüft sie:

```python
if s_alt > 0 and s <= 0:
    print(...)
```

Nur wenn **beides** stimmt, wird die eingerückte Zeile ausgeführt. Sonst wird sie übersprungen.

Führe die Zelle aus und notiere die Zeiten auf dem Arbeitsblatt (Aufgabe 2).

````python
import numpy as np
import matplotlib.pyplot as plt
try:
    from animation import animiere   # für die Animationen (🎬)
except ModuleNotFoundError:          # Notlösung (iPad): Datei direkt von der Webseite laden
    from pyodide.http import open_url
    exec(open_url("https://elmar-siepmann.github.io/bewegungen/files/animation.py").read())

m = 0.2          # Masse in kg
D = 5.0          # Federkonstante in N/m
dt = 0.01        # Zeitschritt in s
N = 600          # 600 Schritte = 6 Sekunden

s = 0.1          # Auslenkung am Anfang in m
v = 0.0          # Geschwindigkeit am Anfang in m/s

t_folge = np.zeros(N+1)
s_folge = np.zeros(N+1)
s_folge[0] = s

for n in range(N):
    F = -D * s
    a = F / m
    v = v + a*dt
    s_alt = s            # alten Ort merken
    s = s + v*dt         # neuer Ort
    t = round((n+1) * dt, 2)
    t_folge[n+1] = t
    s_folge[n+1] = s
    if s_alt > 0 and s <= 0:
        print("nach unten durch die Mitte bei t =", t, "s")

print("größte Auslenkung: ", round(max(s_folge), 3), "m")
print("kleinste Auslenkung:", round(min(s_folge), 3), "m")

plt.plot(t_folge, s_folge)
plt.axhline(0, color="gray")
plt.xlabel("t in s")
plt.ylabel("s in m")
plt.grid()
````
