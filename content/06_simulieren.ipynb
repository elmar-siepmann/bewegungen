# Simulieren: Die Bewegung vorausberechnen

Bisher bist du von **Messdaten** ausgegangen und hast daraus `v`, `a` und schließlich die Kraft `F` berechnet:

s → v → a → F

Jetzt drehst du die Richtung um. Wenn du das **Kraftgesetz** kennst, brauchst du keine Messung mehr. In jedem kleinen Zeitschritt Δt rechnest du

1. die Beschleunigung: a = F / m,
2. die neue Geschwindigkeit: v + a · Δt,
3. den neuen Ort mit dieser neuen Geschwindigkeit: s + v · Δt.

Das sind wieder Summenfolgen, nur rückwärts: F → a → v → s.

Deine Antworten schreibst du auf das Arbeitsblatt.

### ⭐ Zusatz 2: Andersherum?

In allen Schleifen wird zuerst die neue Geschwindigkeit berechnet und dann der Ort mit dieser **neuen** Geschwindigkeit.
Man könnte den Ort auch mit der **alten** Geschwindigkeit berechnen. Dazu vertauschst du die beiden markierten Zeilen.

Führe die Zelle aus, vertausche dann die Zeilen und vergleiche. Probiere auch kleinere Zeitschritte: `dt = 0.025` mit `N = 120` und `dt = 0.01` mit `N = 300`.
(Auf dem Arbeitsblatt: Vertiefung V2)

```python
m = 0.2
D = 5.0
dt = 0.05
N = 60

s = 0.1
v = 0.0

t_folge = np.zeros(N+1)
s_folge = np.zeros(N+1)
s_folge[0] = s

for n in range(N):
    F = -D * s
    a = F / m
    v = v + a*dt         # diese beiden Zeilen
    s = s + v*dt         # vertauschen
    t_folge[n+1] = (n+1) * dt
    s_folge[n+1] = s

print("größte Auslenkung:", max(s_folge), "m")

plt.plot(t_mess, s_mess, "o", label="Messung")
plt.plot(t_folge, s_folge, "-", label="Simulation")
plt.xlabel("t in s")
plt.ylabel("s in m")
plt.legend()
plt.grid()
```
