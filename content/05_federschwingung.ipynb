## Aufgabe 1: Eine Feder schwingt

Ein Wagen (Masse m = 0,2 kg) hängt an einer Feder und rollt ohne Reibung hin und her.
`s` ist die **Auslenkung**: wie weit der Wagen von der Ruhelage entfernt ist (rechts positiv, links negativ).
Alle 0,05 s wurde gemessen, die Daten stehen in `data/feder.csv`.

Die Berechnung von `v` und `a` kennst du aus dem letzten Notebook, sie ist schon vorgegeben.
Führe die Zelle aus und schau dir das Diagramm genau an.

```python
import numpy as np
import matplotlib.pyplot as plt

m = 0.2          # Masse in kg
t, s = np.loadtxt("data/feder.csv", delimiter=",", skiprows=1, unpack=True)

N = len(t)
dt = t[1] - t[0]      # Zeitschritt in s

v = np.zeros(N-1)
for n in range(N-1):
    v[n] = (s[n+1] - s[n]) / dt

a = np.zeros(N-2)
for n in range(N-2):
    a[n] = (v[n+1] - v[n]) / dt

plt.plot(t, s, "o-", label="s in m")
plt.plot(t[:-1], v, "o-", label="v in m/s")
plt.plot(t[:-2], a, "o-", label="a in m/s²")
plt.xlabel("t in s")
plt.legend()
plt.grid()
```
